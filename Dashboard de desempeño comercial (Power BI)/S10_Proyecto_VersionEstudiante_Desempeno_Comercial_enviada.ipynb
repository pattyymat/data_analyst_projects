### Narrativa del Dashboard (Modelo SCQA)

#### 🖥️ Vista General (Overview)

**S (Situación):**  
La empresa Andes Retail Group opera en varios países y comercializa diferentes categorías de productos para distintos segmentos de clientes. La dirección necesita monitorear el desempeño comercial de 2024 y 2025 para comprender cómo evolucionan los ingresos, qué clientes generan mayor valor, qué categorías de productos tienen mayor impacto y qué mercados impulsan los resultados del negocio.

**C (Complicación):**

La visión general evidencia que el desempeño no es uniforme. Se observan diferencias entre:

- La evolución de los ingresos durante el periodo;
- El aporte de los segmentos de clientes;
- El desempeño de las categorías de productos;
- La contribución de los distintos países.

Estas diferencias dificultan identificar si el negocio responde de la misma manera en todos los mercados y para todos los tipos de clientes.

**Q (Pregunta):**   

Ante estas diferencias surge una pregunta natural:

¿Qué factores explican estas variaciones y dónde conviene profundizar el análisis para comprender mejor el desempeño comercial?

**A (Respuesta):**  

La visión general confirma que el desempeño comercial presenta diferencias relevantes entre mercados y segmentos, así como variaciones durante el periodo analizado.

El mayor valor del negocio se concentra principalmente en los segmentos Premium y Estándar y en los mercados de Perú y Chile, aquí es donde conviene profundizar el análisis para priorizar futuras acciones comerciales.

---

#### 🔎 Vista Detalle

**S (Situación):**  

La visión general permitió identificar diferencias entre mercados, segmentos y categorías.

Sin embargo, para orientar decisiones comerciales no basta con conocer dónde existen esas diferencias; también es necesario comprender cómo evolucionan a lo largo del tiempo y qué combinaciones requieren mayor atención.

**C (Complicación):**  

El análisis temporal muestra que los mercados y los segmentos no evolucionan de la misma manera durante el año.

Mientras algunos presentan variaciones más pronunciadas, otros mantienen un comportamiento más estable.

Esto sugiere que aplicar una misma estrategia comercial a todos los mercados y segmentos podría no ser la mejor alternativa.

**Q (Pregunta):**  

A partir de este comportamiento surge una nueva pregunta:

¿Qué patrones temporales ayudan a explicar estas diferencias y en qué mercados o segmentos conviene priorizar el análisis para identificar oportunidades de mejora comercial?

**A (Respuesta):**  

Las principales oportunidades comerciales se concentran en Perú y Chile durante el invierno, donde se observan las mayores disminuciones de ingresos y en los segmentos Premium y Estándar que ofrecen el mayor potencial de gestión por su peso en el negocio y su variabilidad temporal.
